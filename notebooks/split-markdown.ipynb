```python
def pprint(node, depth=0, sep=' ', end='\n', file=None, flush=False):
    print(sep*depth+f"<{node.type} start='{node.start_byte}' end='{node.end_byte}'>", sep=sep, end=end, file=file, flush=flush)
    for child in node.children:
        pprint(child, depth=depth+1, sep=sep, end=end, file=file, flush=flush)
    if not node.children:
        segment = content[node.start_byte:node.end_byte].strip()
        if segment:
            print(sep*(depth+1)+segment.decode("utf-8"), sep=sep, end=end, file=file, flush=flush)
    print(sep*depth+f"</{node.type}>", sep=sep, end=end, file=file, flush=flush)
```
